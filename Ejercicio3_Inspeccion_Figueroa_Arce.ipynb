## Ejercicio 3: Inspección visual de productos

### Ficha PEAS

<table>
<tr>
<th>Componente</th>
<th>Descripción</th>
</tr>

<tr>
<td><strong>Percepción (S)</strong></td>
<td>
<ul>
<li><strong>Matriz de inspección:</strong> matriz de 5×5 capturada por una cámara, donde cada posición representa una zona del producto.</li>
<li><strong>Valor 0:</strong> zona normal o sin defecto.</li>
<li><strong>Valor 1:</strong> zona que presenta un defecto.</li>
<li><strong>Cantidad de defectos:</strong> número total de posiciones con valor 1 en la matriz.</li>
</ul>
</td>
</tr>

<tr>
<td><strong>Entorno (E)</strong></td>
<td>
Sistema automatizado de inspección visual encargado de verificar productos mediante una cámara antes de su aprobación.
</td>
</tr>

<tr>
<td><strong>Acciones (A)</strong></td>
<td>
<ul>
<li><strong>Aprobar:</strong> el producto no presenta defectos.</li>
<li><strong>Enviar a revisión manual:</strong> el producto presenta entre 1 y 3 defectos.</li>
<li><strong>Rechazar:</strong> el producto presenta más de 3 defectos.</li>
</ul>
</td>
</tr>

<tr>
<td><strong>Medida de desempeño (P)</strong></td>
<td>
Clasificar correctamente los productos según la cantidad de defectos detectados, evitando aprobar productos defectuosos y reduciendo revisiones manuales innecesarias.
</td>
</tr>

</table>

### Justificación de las reglas

Las reglas se basan en la cantidad de defectos detectados en la matriz de inspección. Si no se detectan defectos, el producto se aprueba directamente. Cuando se encuentran entre 1 y 3 defectos, el producto se envía a revisión manual para realizar una evaluación adicional. Si se detectan más de 3 defectos, el producto se rechaza debido al mayor nivel de deterioro identificado.
