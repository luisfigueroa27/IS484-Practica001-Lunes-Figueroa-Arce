### Justificación de las reglas

Las reglas priorizan los tickets de urgencia alta para reducir el tiempo de atención de incidentes críticos. Los tickets de urgencia media pueden escalar cuando presentan un tiempo de espera elevado o pertenecen a un cliente premium. Los tickets de urgencia baja se mantienen en el nivel inicial, excepto cuando un cliente premium presenta un tiempo de espera considerable. De esta manera, se combinan urgencia, tiempo de espera y condición del cliente para distribuir adecuadamente la carga de trabajo del equipo de soporte.
