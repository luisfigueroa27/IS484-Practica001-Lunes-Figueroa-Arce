### Justificación de las reglas

Las reglas priorizan los tickets de urgencia alta para reducir el tiempo de atención de incidentes críticos. Los tickets de urgencia media pueden escalar cuando presentan un tiempo de espera elevado o pertenecen a un cliente premium. Los tickets de urgencia baja se mantienen en el nivel inicial, excepto cuando un cliente premium presenta un tiempo de espera considerable. De esta manera, se combinan urgencia, tiempo de espera y condición del cliente para distribuir adecuadamente la carga de trabajo del equipo de soporte.

### Código

La función `agente_soporte` evalúa cada ticket considerando el tiempo de espera, el nivel de urgencia y si el cliente es premium. A partir de estas percepciones, el agente determina si el ticket debe mantenerse en nivel 1 o escalarse a nivel 2 o nivel 3.

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    
    # Normalizar el nivel de urgencia
    nivel_urgencia = nivel_urgencia.lower()

    # Regla 1: cualquier ticket de urgencia alta
    # se escala inmediatamente al nivel 3.
    if nivel_urgencia == "alta":
        return (
            "escalar a nivel 3",
            "urgencia alta reportada por el cliente"
        )

    # Regla 2: urgencia media con espera elevada
    # o cliente premium.
    elif nivel_urgencia == "media":
        
        if tiempo_espera_minutos >= 30 or cliente_premium:
            return (
                "escalar a nivel 2",
                "urgencia media con tiempo de espera elevado "
                "o cliente premium"
            )
        
        else:
            return (
                "mantener en nivel 1",
                "urgencia media sin condiciones de escalamiento"
            )

    # Regla 3: urgencia baja, cliente premium
    # y tiempo de espera elevado.
    elif nivel_urgencia == "baja":
        
        if cliente_premium and tiempo_espera_minutos >= 60:
            return (
                "escalar a nivel 2",
                "cliente premium con tiempo de espera elevado"
            )
        
        else:
            return (
                "mantener en nivel 1",
                "urgencia baja sin condiciones de escalamiento"
            )

    # Control para niveles de urgencia no válidos.
    else:
        return (
            "mantener en nivel 1",
            "nivel de urgencia no reconocido"
        )
```
